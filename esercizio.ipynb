## Esercizio 3 — Profilo accademico dello studente

Partendo dai dati convertiti nell’esercizio precedente, crea un piccolo profilo accademico dello studente.

Usa queste variabili già convertite:

- `eta`
- `iscritto`
- `numero_esami`
- `media_voti`

Crea poi una lista chiamata `voti` contenente almeno 5 voti d’esame.

Calcola e stampa:

- il numero di voti presenti nella lista usando `len()`
- il voto più alto usando `max()`
- il voto più basso usando `min()`
- la somma dei voti usando `sum()`
- la media effettiva dei voti

Poi verifica e stampa il risultato delle seguenti condizioni:

- lo studente è maggiorenne
- lo studente è iscritto
- lo studente ha sostenuto almeno `6` esami
- la media dichiarata è maggiore o uguale a `27`
- la media effettiva calcolata dalla lista è maggiore o uguale a `27`
- la media dichiarata e la media effettiva sono uguali
- lo studente è iscritto e ha sostenuto almeno `6` esami
- lo studente è maggiorenne, iscritto e ha una media effettiva maggiore o uguale a `27`

Non usare ancora `if`, `for` o `while`: limitati a calcolare valori e stampare il risultato delle condizioni.
