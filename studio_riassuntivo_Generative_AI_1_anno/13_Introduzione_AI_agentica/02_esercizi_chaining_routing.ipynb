````python
# ── Imports globali ──────────────────────────────────────────────────────────
import textwrap
from typing import Any

from langchain_ollama import ChatOllama
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# ── Configurazione del modello ───────────────────────────────────────────────
MODEL = "llama3.2"          # Cambia con il modello che hai scaricato su Ollama
BASE_URL = "http://localhost:11434"

llm: Any = ChatOllama(
    model=MODEL,
    base_url=BASE_URL,
    temperature=0.7,
)

# Helper per stampare in modo leggibile
def stampa(titolo: str, testo: str):
    separatore = "─" * 60
    print(f"\n{separatore}")
    print(f"📌 {titolo}")
    print(separatore)
    print(textwrap.fill(testo.strip(), width=70))
    print()

# Verifica connessione a Ollama
test = llm.invoke([HumanMessage(content="Rispondi solo: OK")])
print(f"✅ Ollama risponde correttamente — modello: {MODEL}")
````

```

```

I 3 step della chain sono:

1. **Estrazione degli aspetti** — il modello identifica gli aspetti chiave menzionati nella recensione (es. qualità, prezzo, spedizione, assistenza)
2. **Valutazione del sentiment** — per ciascun aspetto estratto, il modello valuta se il sentiment è positivo, negativo o neutro
3. **Generazione della risposta** — il modello redige una risposta professionale al cliente, tenendo conto del sentiment di ogni aspetto

- Definire i 3 `ChatPromptTemplate` (uno per step)

- Eseguire la pipeline in modo sequenziale, passando l'output di uno step come input del successivo
- Stampare i risultati intermedi e quello finale usando la funzione `stampa()`
